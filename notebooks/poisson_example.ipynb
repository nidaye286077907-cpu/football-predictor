````python
# 2) 训练 Poisson pair 模型（使用示例数据）
from src.poisson import fit_poisson_pair
MODEL_PATH = fit_poisson_pair('data/raw/example_matches.csv')
print('模型已保存：', MODEL_PATH)
````

````python
# 3) 用训练好的模型预测一批赛程的 lambda 与胜平负概率
import pandas as pd
from src.poisson import predict_probs_for_fixtures
fixtures = pd.read_csv('data/raw/example_matches.csv', parse_dates=['date'])[['date','home_team','away_team']].head(10)
df_probs = predict_probs_for_fixtures(MODEL_PATH, fixtures, rho=0.08, max_goals=6)
df_probs
````

````python
# 4) 将预测结果合并赔率并做简单回测示例（调用回测脚本）
from src.utils import merge_odds
from src.backtest import simple_backtest
# 假设我们已经有 predictions.csv（包含 p_H/p_D/p_A 与 odd_h/odd_d/odd_a）
# 下面把 poisson 预测结果写入 predictions.csv，并合并示例赔率文件，然后回测
pred = df_probs.copy()
# 为演示我们把示例赔率合并（示例赔率文件在 data/raw/example_odds.csv）
matches = pd.read_csv('data/raw/example_matches.csv', parse_dates=['date'])
merged = merge_odds(pred.merge(matches[['date','home_team','away_team','home_goals','away_goals']], on=['date','home_team','away_team'], how='left'), 'data/raw/example_odds.csv')
merged.to_csv('data/processed/predictions.csv', index=False)
# 运行回测（默认读取 data/processed/predictions.csv）
hist = simple_backtest('data/processed/predictions.csv', bankroll=1000.0, stake_frac=0.02, min_edge=0.01)
hist.head()
````
